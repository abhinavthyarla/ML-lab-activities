```python
# Airthmetic operators
a = 7
b = 2
print("a + b =",a + b)
print("a - b =",a - b)
print("a * b =",a * b)
print("a / b =",a / b)
print("a ** 2 =",a ** 2)

# Comparison operators
cgpa = 0.4
print("cgpa == 0.4 |", cgpa == 0.4)
print("cgpa > 0.4 |", cgpa > 0.4)
print("cgpa < 0.4 |", cgpa < 0.4)

# logical operators: and / or / not
attendance = 85
placed = (cgpa > 0) and (attendance > 80)
print("placed |", placed)
```

f string

```python
algorithm = "Decision Tree"
score = 0.91578

# Old, clunky way (avoid this):
print("Performance of " + algorithm + " is " + str(score))

# f-string way (use this):
print(f"Performance of {algorithm} is {score}")

# ':.2f' inside the braces rounds a float to 2 decimal places.
print(f"Performance of {algorithm} is {score:.2f}")
```
